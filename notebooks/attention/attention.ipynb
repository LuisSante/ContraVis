```python
import numpy as np
```

##### Tokenización + embeddings
$$
X_{sa} \in \mathbb{R}^{n \times d}
X_{sb} \in \mathbb{R}^{m \times d}
$$

n = number of tokens sa

m = number of tokens sb

d = shape of embedding


```python
sa = "the application is not working"
sb = "the application is working"
```

```python
def tokenize(s):
    return s.split()

def softmax(x):
    e_x = np.exp(x - np.max(x, axis=-1, keepdims=True))
    return e_x / np.sum(e_x, axis=-1, keepdims=True)
```

```python
tok_sa = tokenize(sa)
tok_sb = tokenize(sb)
```

```python
d_model = 8
```

```python
vocab = list(set(tok_sa + tok_sb))
vocab
```

```python
embeddings = {word: np.random.rand(d_model) for word in vocab}
embeddings
```

```python
def embed(tokens):
    return np.array([embeddings[t] for t in tokens])
```

##### Create matrix Q, K, V

$$
W_Q, W_K, W_V \in \mathbb{R}^{d \times d}
$$

$$
Q = X W_Q \in \mathbb{R}^{n \times d}
$$

$$
K = X W_K \in \mathbb{R}^{n \times d}
$$

$$
V = X W_V \in \mathbb{R}^{n \times d}
$$

```python
def create_qkv(X):
    d = X.shape[1]
    Wq = np.random.rand(d, d)
    Wk = np.random.rand(d, d)
    Wv = np.random.rand(d, d)

    Q = X @ Wq
    K = X @ Wk
    V = X @ Wv

    return Q, K, V
```

##### SELF-ATTENTION
$$
Q, K, V \in \mathbb{R}^{n \times d}
$$

$$
S = \frac{Q K^\top}{\sqrt{d}} \in \mathbb{R}^{n \times n}
$$

$$
A = \text{softmax}(S) \in \mathbb{R}^{n \times n}
$$

$$
\text{Output} = A V \in \mathbb{R}^{n \times d}
$$

```python
def self_attention(X):
    Q, K, V = create_qkv(X)

    scores = Q @ K.T / np.sqrt(K.shape[1])
    self_attention_weights = softmax(scores)
    self_attention_output = self_attention_weights @ V

    return self_attention_output, self_attention_weights
```

##### CROSS-ATTENTION

$$
Q \in \mathbb{R}^{n \times d}
$$

$$
K, V \in \mathbb{R}^{m \times d}
$$

$$
S = \frac{Q K^\top}{\sqrt{d}} \in \mathbb{R}^{n \times m}
$$

$$
A = \text{softmax}(S) \in \mathbb{R}^{n \times m}
$$

$$
\text{Output} = A V \in \mathbb{R}^{n \times d}
$$

```python
def cross_attention(X_target, X_candidate):
    d = X_target.shape[1]

    Wq = np.random.rand(d, d)
    Wk = np.random.rand(d, d)
    Wv = np.random.rand(d, d)

    Q = X_target @ Wq
    K = X_candidate @ Wk
    V = X_candidate @ Wv

    scores = Q @ K.T / np.sqrt(K.shape[1])
    cross_attention_weights = softmax(scores)
    cross_attention_output = cross_attention_weights @ V

    return cross_attention_output, cross_attention_weights
```

##### TEST

```python
X_sa = embed(tok_sa)
X_sb = embed(tok_sb)

# self-attention
self_attention_output, self_attention_weights = self_attention(X_sa)

# cross-attention
cross_attention_, cross_attention_weights = cross_attention(X_sa, X_sb)

print("Self-attention shape:", self_attention_weights.shape)
print("Cross-attention shape:", cross_attention_weights.shape)
```
