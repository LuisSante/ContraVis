```python
from pathlib import Path
import os

text_root = Path("../infra/CUAD_v1/full_contract_txt/")

txt_files = [p for p in text_root.rglob("*") if p.is_file() and p.suffix.lower() == ".txt"]

num_tokens_list = []
for txt_path in txt_files:
    with open(txt_path, "r", encoding="utf-8") as f:
        text = f.read()
        tokens = text.split()
        num_tokens = len(tokens)

        only_name_txt = txt_path.stem
        num_tokens_list.append(num_tokens)

min_value = min(num_tokens_list)
max_value = max(num_tokens_list)
avg_value = sum(num_tokens_list) / len(num_tokens_list)
sum_value = sum(num_tokens_list)

print(f"Min: {min_value} tokens")
print(f"Max: {max_value} tokens")
print(f"Avg: {avg_value} tokens")
print(f"Sum: {sum_value} tokens")
```
