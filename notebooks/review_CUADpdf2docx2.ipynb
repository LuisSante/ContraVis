```python
from pdf2docx import Converter
from pathlib import Path
import os

pdf_root = Path("../infra/CUAD_v1/full_contract_pdf/")
docx_root = Path("../infra/CUAD_v1/full_contract_docx/")

pdf_files = [p for p in pdf_root.rglob("*") if p.is_file() and p.suffix.lower() == ".pdf"]

print(f"Find {len(pdf_files)} PDFs")

for pdf_path in pdf_files:
    try:
        rel_path = pdf_path.relative_to(pdf_root)
        docx_path = docx_root / rel_path.with_suffix(".docx")

        docx_path.parent.mkdir(parents=True, exist_ok=True)

        cv = Converter(str(pdf_path))
        cv.convert(str(docx_path), start=0, end=None)
        cv.close()

        print(f"OK -> {pdf_path}")

    except Exception as e:
        print(f"FAIL -> {pdf_path} | {e}")
```
