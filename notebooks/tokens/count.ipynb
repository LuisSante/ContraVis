```python
import json
import tiktoken
```

```python
infra_path = "../../infra/"
json_path = f"{infra_path}json/"
cuad_json = f"{infra_path}CUAD_v1/CUAD_v1.json"
cuad_json, json_path
```

```python
with open(cuad_json, "r") as f:
    cuad_data = json.load(f)
```

```python
import re 

def clean_extreme_whitespace(text: str):
    text = re.sub(r'\s+', ' ', text)
    return text.strip()

document_list = [
  "BELLICUMPHARMACEUTICALS,INC_05_07_2019-EX-10.1-Supply Agreement",
  "DigitalCinemaDestinationsCorp_20111220_S-1_EX-10.10_7346719_EX-10.10_Affiliate Agreement",
  "MusclepharmCorp_20170208_10-KA_EX-10.38_9893581_EX-10.38_Co-Branding Agreement",
  "ArcaUsTreasuryFund_20200207_N-2_EX-99.K5_11971930_EX-99.K5_Development Agreement",
  "TomOnlineInc_20060501_20-F_EX-4.46_749700_EX-4.46_Co-Branding Agreement",
]

data = cuad_data["data"]

new_json_only_context = []

COST_INPUT_PER_MILLION = 0.40  
COST_OUTPUT_PER_MILLION = 1.60  
cost_input_per_token = COST_INPUT_PER_MILLION / 1_000_000
enc = tiktoken.encoding_for_model("gpt-4.1")

for index, item in enumerate(cuad_data['data'], start=1):
    title = item['title']

    if title in document_list:
        original_context = item['paragraphs'][0]['context']
        
        clean_text = " ".join(original_context.split())
        
        tokens = enc.encode(clean_text)
        number_tokens = len(tokens)

        new_json_only_context.append({
            "id": f"CUAD_{index}",
            "title": title,
            "context": clean_text,
            "total_number_tokens": number_tokens,
            "cost_document_input": number_tokens * cost_input_per_token,
        })

new_json_only_context
```
